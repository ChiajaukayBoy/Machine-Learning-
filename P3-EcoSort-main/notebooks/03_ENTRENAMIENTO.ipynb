## Funciones auxiliares

```python
def calcular_metricas(y_true, y_pred, y_pred_proba=None):
    f1_macro = f1_score(y_true, y_pred, average='macro')
    f1_micro = f1_score(y_true, y_pred, average='micro')

    auc_pr = None
    if y_pred_proba is not None:
        y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
        auc_pr_scores = []
        for i in range(num_clases):
            precision, recall, _ = precision_recall_curve(y_true_bin[:, i], y_pred_proba[:, i])
            auc_pr_scores.append(auc(recall, precision))
        auc_pr = np.mean(auc_pr_scores)

    return {
        'f1_macro': f1_macro,
        'f1_micro': f1_micro,
        'auc_pr': auc_pr
    }

def graficar_curva_perdida(historial_train, historial_val, titulo, nombre_archivo):
    fig, ax = plt.subplots(figsize=(10, 6))
    epochs = range(1, len(historial_train) + 1)

    ax.plot(epochs, historial_train, 'o-', label='Train Loss', linewidth=2, markersize=5, color='steelblue')
    ax.plot(epochs, historial_val, 's-', label='Val Loss', linewidth=2, markersize=5, color='coral')

    ax.set_xlabel('Epoch', fontsize=12, fontweight='bold')
    ax.set_ylabel('Loss', fontsize=12, fontweight='bold')
    ax.set_title(titulo, fontsize=14, fontweight='bold')
    ax.legend(fontsize=11)
    ax.grid(alpha=0.3)

    plt.tight_layout()
    plt.savefig(ruta_figuras / nombre_archivo, format='svg', bbox_inches='tight')
    plt.show()

def graficar_matriz_confusion(y_true, y_pred, titulo, nombre_archivo):
    cm = confusion_matrix(y_true, y_pred)
    cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(cm_norm, annot=True, fmt='.3f', cmap='Blues', xticklabels=clases, yticklabels=clases, ax=ax, cbar_kws={'label': 'Ratio'})
    ax.set_xlabel('Predicted Label', fontsize=12, fontweight='bold')
    ax.set_ylabel('True Label', fontsize=12, fontweight='bold')
    ax.set_title(titulo, fontsize=14, fontweight='bold')

    plt.tight_layout()
    plt.savefig(ruta_figuras / nombre_archivo, format='svg', bbox_inches='tight')
    plt.show()
```

## 1. Modelo Baseline: Logistic Regression

### 1.1 Optimización de hiperparámetros

```python
param_grid_lr_autoencoder = {
    'C': [ 1],
    'penalty': ['l2'],
    'solver': ['saga'],
    'max_iter': [2000]
}

print("Configuración de hiperparámetros para Logistic Regression con Autoencoders:")
for param, valores in param_grid_lr_autoencoder.items():
    print(f"  {param}: {valores}")

resultados_comparacion_lr = {}

features_dict_autoencoders = {
    #'Autoencoder Variacional': {
    #    'train': features_train_autoencoder_variational,
    #    'val': features_val_autoencoder_variational
    #},
    'Convolutional Autoencoder': {
        'train': features_train_autoencoder_convolutional,
        'val': features_val_autoencoder_convolutional
    }
}

for nombre_modelo, features in features_dict_autoencoders.items():
    print(f"\n{'='*50}")
    print(f"Procesando modelo: {nombre_modelo}")
    print(f"{'='*50}")

    # Extraer features
    X_train = features['train']
    X_val = features['val']

    print(f"Dimensiones - Train: {X_train.shape}, Val: {X_val.shape}")

    # Búsqueda de hiperparámetros con Halving Grid Search
    print("Iniciando Halving Grid Search...")
    modelo_lr = LogisticRegression(random_state=42, multi_class='multinomial')
    halving_search_lr = HalvingGridSearchCV(
        modelo_lr,
        param_grid_lr_autoencoder,
        factor=3,  # Reduce a 1/3 en cada iteración
        cv=3,
        scoring='f1_macro',
        n_jobs=1,  # Usar todos los cores disponibles
        random_state=42,
        verbose=3,
        min_resources='exhaust'
        
    )

    tiempo_inicio = time.time()
    halving_search_lr.fit(X_train, y_train)
    tiempo_total_lr = time.time() - tiempo_inicio

    mejor_modelo_lr = halving_search_lr.best_estimator_
    mejores_params_lr_ae = halving_search_lr.best_params_

    print(f"Mejores parámetros: {mejores_params_lr_ae}")
    print(f"Tiempo de entrenamiento: {tiempo_total_lr:.2f} segundos")
    print(f"Mejor score: {halving_search_lr.best_score_:.4f}")

    # Generar historial de entrenamiento (simulación)
    print("Generando historial de entrenamiento...")
    historial_train_lr = []
    historial_val_lr = []

    for i in range(1, 1):
        modelo_temp = LogisticRegression(**mejores_params_lr_ae, random_state=42, multi_class='multinomial')
        modelo_temp.fit(X_train, y_train)

        y_pred_train_temp = modelo_temp.predict(X_train)
        y_pred_val_temp = modelo_temp.predict(X_val)

        y_pred_train_proba_temp = modelo_temp.predict_proba(X_train)
        y_pred_val_proba_temp = modelo_temp.predict_proba(X_val)

        loss_train = log_loss(y_train, y_pred_train_proba_temp)
        loss_val = log_loss(y_val, y_pred_val_proba_temp)

        historial_train_lr.append(loss_train)
        historial_val_lr.append(loss_val)

    # Predicciones finales
    y_pred_train_lr = mejor_modelo_lr.predict(X_train)
    y_pred_val_lr = mejor_modelo_lr.predict(X_val)
    y_pred_proba_val_lr = mejor_modelo_lr.predict_proba(X_val)

    # Calcular métricas
    metricas_lr = calcular_metricas(y_val, y_pred_val_lr, y_pred_proba_val_lr)

    # Almacenar resultados
    resultados_lr = pd.DataFrame({
        'Métrica': ['F1 Macro', 'F1 Micro', 'AUC-PR', 'Tiempo entrenamiento (s)'],
        'Valor': [
            f"{metricas_lr['f1_macro']:.4f}",
            f"{metricas_lr['f1_micro']:.4f}",
            f"{metricas_lr['auc_pr']:.4f}",
            f"{tiempo_total_lr:.2f}"
        ]
    })

    print(f"\nResultados para {nombre_modelo}:")
    print(resultados_lr.to_string(index=False))

    # Guardar resultados para comparación
    resultados_comparacion_lr[nombre_modelo] = {
        'metricas': metricas_lr,
        'tiempo': tiempo_total_lr,
        'mejores_params': mejores_params_lr_ae,
        'historial_train': historial_train_lr,
        'historial_val': historial_val_lr,
        'dataframe': resultados_lr
    }

    # Generar gráficos
    nombre_archivo_clean = nombre_modelo.lower().replace(' ', '_').replace('-', '_')

    print(f"Generando gráficos para {nombre_modelo}...")
    graficar_matriz_confusion(
        y_val,
        y_pred_val_lr,
        f'Logistic Regression ({nombre_modelo})-Confusion Matrix',
        f'03_train_02_lr_confusion_matrix_{nombre_archivo_clean}.svg'
    )

    graficar_curva_perdida(
        historial_train_lr,
        historial_val_lr,
        f'Curva de pérdida - Logistic Regression ({nombre_modelo})',
        f'03_train_01_lr_loss_curve_{nombre_archivo_clean}.svg'
    )

    print(f"Procesamiento completado para {nombre_modelo}")

# %% Comparación de resultados entre modelos de LR
print(f"\n{'='*60}")
print("COMPARACIÓN DE RESULTADOS - LOGISTIC REGRESSION CON AUTOENCODERS")
print(f"{'='*60}")

# Crear DataFrame de comparación
comparison_data_lr = []
for modelo, resultados in resultados_comparacion_lr.items():
    comparison_data_lr.append({
        'Modelo': modelo,
        'F1 Macro': resultados['metricas']['f1_macro'],
        'F1 Micro': resultados['metricas']['f1_micro'],
        'AUC-PR': resultados['metricas']['auc_pr'],
        'Tiempo (s)': resultados['tiempo']
    })

df_comparacion_lr = pd.DataFrame(comparison_data_lr)
df_comparacion_lr = df_comparacion_lr.sort_values('F1 Macro', ascending=False)

print("\nRanking de modelos por F1 Macro:")
print(df_comparacion_lr.to_string(index=False))

# %% Gráfico de comparación
plt.figure(figsize=(12, 8))

# Subplot 1: F1 Scores
plt.subplot(2, 2, 1)
x_pos = np.arange(len(df_comparacion_lr))
plt.bar(x_pos - 0.2, df_comparacion_lr['F1 Macro'], 0.4, label='F1 Macro', alpha=0.8)
plt.bar(x_pos + 0.2, df_comparacion_lr['F1 Micro'], 0.4, label='F1 Micro', alpha=0.8)
plt.xlabel('Modelos')
plt.ylabel('F1 Score')
plt.title('Comparación F1 Scores - LR con Autoencoders')
plt.xticks(x_pos, df_comparacion_lr['Modelo'], rotation=45, ha='right')
plt.legend()
plt.grid(True, alpha=0.3)

# Subplot 2: AUC-PR
plt.subplot(2, 2, 2)
plt.bar(x_pos, df_comparacion_lr['AUC-PR'], alpha=0.8, color='green')
plt.xlabel('Modelos')
plt.ylabel('AUC-PR')
plt.title('Comparación AUC-PR - LR con Autoencoders')
plt.xticks(x_pos, df_comparacion_lr['Modelo'], rotation=45, ha='right')
plt.grid(True, alpha=0.3)

# Subplot 3: Tiempo de entrenamiento
plt.subplot(2, 2, 3)
plt.bar(x_pos, df_comparacion_lr['Tiempo (s)'], alpha=0.8, color='orange')
plt.xlabel('Modelos')
plt.ylabel('Tiempo (segundos)')
plt.title('Tiempo de Entrenamiento - LR con Autoencoders')
plt.xticks(x_pos, df_comparacion_lr['Modelo'], rotation=45, ha='right')
plt.grid(True, alpha=0.3)

# Subplot 4: Curvas de pérdida comparadas
plt.subplot(2, 2, 4)
for modelo, resultados in resultados_comparacion_lr.items():
    plt.plot(resultados['historial_val'], label=f'{modelo}', marker='o', alpha=0.7)
plt.xlabel('Iteración')
plt.ylabel('Pérdida de Validación')
plt.title('Comparación Curvas de Pérdida - LR')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(ruta_figuras / 'comparacion_lr_autoencoders.svg', format='svg', bbox_inches='tight')
plt.show()

# %% Resumen final LR
print(f"\n{'='*60}")
print("RESUMEN FINAL - LOGISTIC REGRESSION")
print(f"{'='*60}")

mejor_modelo_lr = df_comparacion_lr.iloc[0]['Modelo']
mejor_f1_macro_lr = df_comparacion_lr.iloc[0]['F1 Macro']

print(f"Mejor modelo LR: {mejor_modelo_lr}")
print(f"F1 Macro: {mejor_f1_macro_lr:.4f}")
print(f"\nMejores parámetros del modelo ganador:")
for param, valor in resultados_comparacion_lr[mejor_modelo_lr]['mejores_params'].items():
    print(f"  {param}: {valor}")
```

### 3.1 Definición de arquitectura CNN

```python
class CNN_Clasificador(nn.Module):
    def __init__(self, num_clases, dropout_rate=0.5):
        super(CNN_Clasificador, self).__init__()

        self.capa_conv1 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Dropout2d(dropout_rate * 0.5)
        )

        self.capa_conv2 = nn.Sequential(
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Dropout2d(dropout_rate * 0.6)
        )

        self.capa_conv3 = nn.Sequential(
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),
            nn.Dropout2d(dropout_rate * 0.7)
        )

        self.capa_fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 16 * 16, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(dropout_rate),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(dropout_rate),
            nn.Linear(128, num_clases)
        )

    def forward(self, x):
        x = self.capa_conv1(x)
        x = self.capa_conv2(x)
        x = self.capa_conv3(x)
        x = self.capa_fc(x)
        return x
```

### 3.3 Función de entrenamiento CNN

```python
def entrenar_cnn(modelo, train_loader, val_loader, criterio, optimizador, num_epochs):
    historial_train_loss = []
    historial_val_loss = []
    mejor_val_loss = float('inf')

    for epoch in range(num_epochs):
        modelo.train()
        train_loss = 0.0

        for imagenes, etiquetas in tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs}'):
            imagenes, etiquetas = imagenes.to(device), etiquetas.to(device)

            optimizador.zero_grad()
            salidas = modelo(imagenes)
            loss = criterio(salidas, etiquetas)
            loss.backward()
            optimizador.step()

            train_loss += loss.item() * imagenes.size(0)

        train_loss /= len(train_loader.dataset)

        modelo.eval()
        val_loss = 0.0
        with torch.no_grad():
            for imagenes, etiquetas in val_loader:
                imagenes, etiquetas = imagenes.to(device), etiquetas.to(device)
                salidas = modelo(imagenes)
                loss = criterio(salidas, etiquetas)
                val_loss += loss.item() * imagenes.size(0)

        val_loss /= len(val_loader.dataset)

        historial_train_loss.append(train_loss)
        historial_val_loss.append(val_loss)

        if val_loss < mejor_val_loss:
            mejor_val_loss = val_loss

    return historial_train_loss, historial_val_loss
```

### 3.4 Búsqueda de hiperparámetros CNN

```python
configuraciones_cnn = [
    {'batch_size': 32, 'lr': 0.001, 'dropout': 0.3, 'weight_decay': 0.0001},
    {'batch_size': 64, 'lr': 0.001, 'dropout': 0.4, 'weight_decay': 0.0001},
    {'batch_size': 32, 'lr': 0.0005, 'dropout': 0.5, 'weight_decay': 0.0005},
    {'batch_size': 64, 'lr': 0.0005, 'dropout': 0.4, 'weight_decay': 0.0001},
    {'batch_size': 32, 'lr': 0.0001, 'dropout': 0.5, 'weight_decay': 0.001}
]

mejor_f1_cnn = 0
mejores_params_cnn = None
mejor_modelo_cnn = None
mejor_historial_cnn = None

resultados_busqueda_cnn = []

for idx, config in enumerate(configuraciones_cnn):
    train_loader = DataLoader(dataset_train, batch_size=config['batch_size'], shuffle=True)
    val_loader = DataLoader(dataset_val, batch_size=config['batch_size'], shuffle=False)

    modelo = CNN_Clasificador(num_clases, dropout_rate=config['dropout']).to(device)
    criterio = nn.CrossEntropyLoss()
    optimizador = optim.Adam(modelo.parameters(), lr=config['lr'], weight_decay=config['weight_decay'])

    historial_train, historial_val = entrenar_cnn(modelo, train_loader, val_loader, criterio, optimizador, num_epochs=20)

    modelo.eval()
    y_pred_val_cnn = []
    y_pred_proba_val_cnn = []
    with torch.no_grad():
        for imagenes, _ in val_loader:
            imagenes = imagenes.to(device)
            salidas = modelo(imagenes)
            probabilidades = torch.softmax(salidas, dim=1)
            _, predicciones = torch.max(salidas, 1)
            y_pred_val_cnn.extend(predicciones.cpu().numpy())
            y_pred_proba_val_cnn.extend(probabilidades.cpu().numpy())

    f1 = f1_score(y_val, y_pred_val_cnn, average='macro')
    resultados_busqueda_cnn.append({'Config': idx+1, 'F1_Macro': f1, **config})

    if f1 > mejor_f1_cnn:
        mejor_f1_cnn = f1
        mejores_params_cnn = config
        mejor_modelo_cnn = modelo
        mejor_historial_cnn = (historial_train, historial_val)
```

```python
import optuna
def crear_modelo(trial):
    num_capas = trial.suggest_int('num_capas', 1, 10)
    dropout = trial.suggest_float('dropout', 0.2, 0.5)
    filtros_base = trial.suggest_categorical('filtros_base', [16, 32, 64])

    capas = []
    in_channels = 3
    size = 128

    for i in range(num_capas):
        out_channels = filtros_base * (2 ** i)
        capas.append(nn.Conv2d(in_channels, out_channels, 3, padding=1))
        capas.append(nn.BatchNorm2d(out_channels))
        capas.append(nn.ReLU())
        capas.append(nn.MaxPool2d(2, 2))
        capas.append(nn.Dropout2d(dropout * (0.5 + i * 0.1)))
        in_channels = out_channels
        size = size // 2

    return nn.Sequential(
        *capas,
        nn.Flatten(),
        nn.Linear(in_channels * size * size, 128),
        nn.ReLU(),
        nn.Dropout(dropout),
        nn.Linear(128, num_clases)
    )

def objetivo(trial):
    modelo = crear_modelo(trial).to(device)
    lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
    batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])

    train_loader = DataLoader(dataset_train, batch_size=batch_size, shuffle=True)
    val_loader = DataLoader(dataset_val, batch_size=batch_size)

    # Class weights para desbalance
    class_weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
    criterio = nn.CrossEntropyLoss(weight=torch.FloatTensor(class_weights).to(device))
    optimizador = optim.Adam(modelo.parameters(), lr=lr)

    # Entrenar 10 epochs
    for epoch in range(10):
        modelo.train()
        for imgs, labels in train_loader:
            imgs, labels = imgs.to(device), labels.to(device)
            optimizador.zero_grad()
            loss = criterio(modelo(imgs), labels)
            loss.backward()
            optimizador.step()

    # Evaluar
    modelo.eval()
    preds = []
    with torch.no_grad():
        for imgs, _ in val_loader:
            preds.extend(torch.argmax(modelo(imgs.to(device)), 1).cpu().numpy())

    return f1_score(y_val, preds, average='macro')

# Ejecutar búsqueda
study = optuna.create_study(direction='maximize')
study.optimize(objetivo, n_trials=30)

print(f"Mejores parámetros: {study.best_params}")
print(f"Mejor F1: {study.best_value:.4f}")
```
